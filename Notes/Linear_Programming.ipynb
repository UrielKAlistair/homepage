# Linear Programming

Linear optimization problems are ubiquitous. Such a problem, where the objective is simply to minimise a certain linear function subject a set of linear constraints, is called a linear programming problem.

Solving these first proved to be of great value during WW II both for the Societs and the US Air force, and since then it has been a powerful tool in concerns of logistics management.

## Standard form

Typically the problem looks like this:
$$
\min \; c^T x \\
\text{s.t.}\; Ax \ge b \\
x\ge 0.
$$

- If the problem is $\max$, then multiply $c$ by $-1$ and convert to a minimisation problem.

- If there exists any constraint $a^Tx \le b_i$, then multiply by $-1$ to get $(-a)^Tx \ge -b_i$.

- If there exists any equality constraint $a^Tx = b_i$, note that it is possible to simply state it as $a^Tx \ge b_i$ and $a^Tx \le b_i$ simultaneously, the latter of which can be made into a $\ge$ constraint as per above.

- Replace any non-positive variables ($x_i$) with equivalent non-negative ones ($-x_i$) and change the sign in constraints accordingly.

- Replace any free varaibles (that is, no non-positivity/non-negativity constraint) with two variables: $x_i = x_i^+ - x_i^-$ both of which are non-negative.

After all this, we ensure that any general LPP can be brought into this standard minimsation form.

## Certificate for Optimality and the Dual

These set of equations have infinitely many *feasible* solutions. Amongst these, it is our interest to find the solution with the smallest objective function. Let's say you found an algorithm which solves this; how can you convince me that your answer is in fact the optimum value?

The answer is that you prove a lower bound on the objective function. For any feasible value of $x$, $c^Tx \ge \ell$. If you can show a feasible $\bar{x}$ such that $c^T\bar{x} = \ell$, then it ***has*** to be the optimum.

That is, we want the strictest possible lower bound so that your optimum can actually go and meet it; if you proved a loose bound, even lower than the global optimum, it would be impossible for you to get $c^T\bar{x} = \ell$.

We want  $$c^Tx \ge \ell$$

We have
$$Ax \ge b$$

which is mighty close. We have full system of inequalities in $x_i$, and we just want one single inequality that, in some as of now unclear way, is related to $c^T$. We just take linear combinations of these inequalities that we have by pre-multiplying with a row vector:

$$y^T Ax \ge y^Tb$$

(In this process, no term of $y^T$ can be negative, or it would flip the inequality.)

The only thing we need is

$$ c^T \ge y^TA $$

then

$$c^Tx \ge y^TAx \ge y^Tb$$

making $y^Tb$ the lower bound on the objective function we seek.

In summary, For any $y$ that satisfies $ y^TA \le c^T \;; y \ge 0 $, $y^Tb$ is a lower bound on our objective function. We want the highest lower bound possible:

$$
\max \; y^T b \\
\text{s.t.}\; y^TA \le c^T \\
y\ge 0.
$$

Transposing to remove the transpose on $y$:

$$
\max \; b^Ty \\
\text{s.t.}\; A^Ty \le c \\
y\ge 0.
$$

This is known as the "Dual" problem to the "Primal" LP that we started with:

$$
\min \; c^T x \\
\text{s.t.}\; Ax \ge b \\
x\ge 0.
$$

By design, this set of systems satisfy that any solution to the dual lower bounds all solutions to the primal. This is called "Weak Duality".

The incredible thing is that the theorem of "Strong duality" for LPPs tells us that with a guarantee, both their optima are the same. The best lower bound we can find for $c^Tx$ this way always reaches the global optimum.

### General Dual form

Although I first showed that all LPs could be put into the standard form and then stated duality, this can be done for any LPP you encounter in the wild:

Now, $y_i$ are simply the multiplier to the rows in the above matrix. For a maximisation problem, we seek an _upper_ bound, so we want a $\le$ constraint. Thus, $y_1$ needs to be negative to cause that. $y_2$ and $y_4$ need to be positive to retain the current inequality.

$y_3$ can take literally any value since it won't affect the equality.

Previously, the sum of the rows had to add up something elementwise less than $c^T$, so that we could argue that since $x$ was non-negative, it was a lower bound. ($4x$ is less than $5x$; $4x$ gives a lower bound as $x$ is non-negative)

If a variable were free, then we must exactly match $c^T$, as we cannot argue like above.

Non-positive variable -> Greater value than $c^T$ get a lower bound.

For a maximisation problem, to seek an upper bound, the alternate needs to be argued.

### Summary



Here's a quick memorization tool (Which we've ended up intuitively arguing for):

Natural things stay natural.
For variables, being non-negative is considered natural.

For a max problem, a $\le$ inequality is natural. This becomes a non-negative variable. $\ge$ is unnatural so it becomes non-positive. Equality always becomes a free variable.

min, $\ge$ is natural becomes a non-negative variable; $\le$ becomes non-positive.

A non-negative variable corresponds to a natural constraint, which is $\le$ for max and $\ge$ for min. A non-positve variable corresponds to a $\le$ constraint for min and $\ge$ for max.

However, hilariously, the way I discovered the dual above, I actually found the kernel of the CS conditions even earlier:

$$c^Tx \ge y^TAx \ge y^Tb$$

which, due to strong duality, gives at the optima:

$$c^T\bar{x} = \bar{y}^TA\bar{x} = \bar{y}^Tb$$

$$ \bar{y}^TA\bar{x} = \bar{y}^Tb \\
  \bar{y}^T (A\bar{x} -b) = 0 \\
$$

Since we are working with the standard minimsation form $Ax \ge b$, we have $y \ge 0$. We also have $Ax -b \ge 0$ For non negative vectors to dot product to zero, at each component, either $a_i^Tx = b$ or $y_i=0$.

 - For each primal constraint, either it holds with equality, or the corresponding dual variable is zero.


$$ \bar{y}^TA\bar{x} = c^T\bar{x} \\
  (c^T - \bar{y}^T A)\bar{x} = 0 \\
$$


- For each dual constraint, either it holds with equality, or the corresponding primal variable is zero.

# Convex Duality

It's worth mentioning that duality is much broader than just LPPs:

Given

$$\min_x f(x) \\
\textit{s.t } \; g_i(x) \le 0 \;; i \in [1..n]$$

We again seek a lower bound for the objective function, which we can then maximise. Here, the constraints are all negative, so adding any linear combination of them to the objective results in a smaller value than $f(x)$ for all _feasible_ points:

$$\mathcal{L}(x,\alpha) :=  f(x) + \sum_{i=1}^n \alpha_i \cdot g_i(x)$$

However, this is not a lower bound for $f(x)$ as such, for _all_ $x$. That still needs a min over x, which makes this step feel a bit silly.
